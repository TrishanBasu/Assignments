```python
#Create a sementic network with visualization using python

import networkx as nx
import matplotlib.pyplot as plt

G = nx.DiGraph()

# relationships
relationships = [
    ("Football", "Players"),
    ("Football", "Teams"),
    ("Football", "Matches"),
    ("Players", "Goals"),
    ("Teams", "Matches"),
    ("Matches", "Referee"),
]

G.add_edges_from(relationships)

pos = nx.spring_layout(G, seed=42, k=1.5)

nx.draw_networkx_nodes(
    G,
    pos,
    node_size=3000,
    node_color="lightblue",
    edgecolors="black"
)

nx.draw_networkx_edges(
    G,
    pos,
    arrows=True,
    arrowsize=20,
    edge_color="gray"
)

nx.draw_networkx_labels(
    G,
    pos,
    font_size=9,
    font_weight="bold"
)


plt.title("Semantic Network - Football")
plt.axis("off")
plt.tight_layout()
plt.show()
```
